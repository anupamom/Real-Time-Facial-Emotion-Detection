```python

import cv2
from keras.models import model_from_json
import numpy as np
import time

# Load the model
json_file = open("emotiondetector.json", "r")
model_json = json_file.read()
json_file.close()
model = model_from_json(model_json)
model.load_weights("emotiondetector.h5")

# Load Haar Cascade
haar_file = cv2.data.haarcascades + 'haarcascade_frontalface_default.xml'
face_cascade = cv2.CascadeClassifier(haar_file)

# Emotion labels
labels = {
    0: 'Angry',
    1: 'Disgust',
    2: 'Fear',
    3: 'Happy',
    4: 'Neutral',
    5: 'Sad',
    6: 'Surprise'
}

# Emotion colors (RGB)
emotion_colors = {
    'Angry': (0, 0, 255),
    'Disgust': (0, 255, 0),
    'Fear': (128, 0, 128),
    'Happy': (0, 255, 255),
    'Neutral': (255, 255, 255),
    'Sad': (255, 0, 0),
    'Surprise': (255, 255, 0)
}

# Preprocess image
def extract_features(image):
    image = np.array(image)
    image = image.reshape(1, 48, 48, 1)
    return image / 255.0

# Start camera
webcam = cv2.VideoCapture(0)

# Set the camera resolution (bigger)
webcam.set(cv2.CAP_PROP_FRAME_WIDTH, 1280)   # Width
webcam.set(cv2.CAP_PROP_FRAME_HEIGHT, 720)   # Height

font = cv2.FONT_HERSHEY_SIMPLEX
prev_time = time.time()

while True:
    ret, frame = webcam.read()
    if not ret:
        print("Failed to grab frame")
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    faces = face_cascade.detectMultiScale(gray, 1.3, 5)

    for (x, y, w, h) in faces:
        face = gray[y:y+h, x:x+w]
        face = cv2.resize(face, (48, 48))
        face_features = extract_features(face)
        prediction = model.predict(face_features)
        emotion = labels[int(np.argmax(prediction))]

        color = emotion_colors.get(emotion, (255, 255, 255))

        # Draw rectangle around face
        cv2.rectangle(frame, (x, y), (x+w, y+h), color, 2)
        cv2.putText(frame, emotion, (x, y-10), font, 0.9, color, 2, cv2.LINE_AA)

    # FPS Counter (optional)
    current_time = time.time()
    fps = int(1 / (current_time - prev_time))
    prev_time = current_time
    cv2.putText(frame, f'FPS: {fps}', (frame.shape[1] - 100, 30), font, 0.6, (0, 255, 0), 2)

    # Display result (full screen mode)
    cv2.namedWindow("Emotion Detection", cv2.WINDOW_NORMAL)
    cv2.setWindowProperty("Emotion Detection", cv2.WND_PROP_FULLSCREEN, cv2.WINDOW_FULLSCREEN)
    cv2.imshow("Emotion Detection", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

webcam.release()
cv2.destroyAllWindows()
```
